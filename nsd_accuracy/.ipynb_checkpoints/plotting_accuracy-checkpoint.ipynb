```python

import pandas as pd
df = pd.read_csv("NSD_accuracy.csv")
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# clean data
df = df.drop(columns=["Unnamed: 0"], errors="ignore")
df = df[df["recall_condition"] != "full"]  # Exclude 'full' condition

# Use only 3 colors (one per remaining condition)
custom_palette = ["#66c2a5", "#fc8d62", "#8da0cb"]  # green-orange-purple combo

# Set style
sns.set(style="whitegrid")

# Plot
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x="recall_condition",
    y="accuracy",
    hue="recall_condition",  # Fixes the FutureWarning
    errorbar='sd',
    palette=custom_palette,
    legend=False  # Hide duplicate legend
)

# Customize appearance
plt.title("Accuracy by Recall Condition", fontsize=14)
plt.xlabel("Recall Condition", fontsize=12)
plt.ylabel("Mean Accuracy", fontsize=12)
plt.ylim(0, 1)
plt.tight_layout()

# Save high-res figure
plt.savefig("accuracy_by_condition.png", dpi=300)

# Show the plot
plt.show()
```
